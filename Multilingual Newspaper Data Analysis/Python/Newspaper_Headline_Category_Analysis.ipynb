```python
print("Total records:", len(df))
```

```python
print(df["Newspaper"].value_counts())
```

```python
print(df["Category"].value_counts())
```

```python
df["Headline_Length"] = df["Headline"].str.len()
print("Average Headline Length:", df["Headline_Length"].mean())
```

```python
avg_length = df.groupby("Newspaper")["Headline_Length"].mean()
print(avg_length.sort_values(ascending=False))
```

```python
print(df["Is_Advertisement"].value_counts())
```

```python
df['Newspaper'].value_counts().plot(kind="bar")

plt.title("Number of Headlines by Newspaper")
plt.xlabel("Newspaper")
plt.ylabel("Number of Headlines")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.hist(df["Headline_Length"], bins=10)

plt.title("Distribution of Headline Length")
plt.xlabel("Headline Length")
plt.ylabel("Number of Headlines")
plt.show()
```
